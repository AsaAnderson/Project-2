**2. Basic Data Exploration**

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("metro-hoops.csv")

# Display the first five rows
print("First five rows:")
display(df.head())

# Display the last five rows
print("Last five rows:")
display(df.tail())

# Display the number of rows and columns
print("Dataset dimensions:", df.shape)

# Display column names
print("Column names:")
print(df.columns.tolist())
```

**3. Data Types and Structures**

```python
# Display information about the dataset
df.info()

# Identify numerical columns
print("Numerical columns:")
print(df.select_dtypes(include="number").columns.tolist())

# Identify categorical/object columns
print("\nCategorical columns:")
print(df.select_dtypes(include="object").columns.tolist())
```

**5. Missing or Duplicate Data**

```python
print("\nTotal missing values:")
print(df.isnull().sum().sum())
```

```python
print("\nNumber of duplicate rows:")
print(df.duplicated().sum())
```

**6. Add a New Column Using Existing Columns**

```python


df["estimated_total_points"] = (
    df["points_per_game"] * df["games"]
)


print(df[[
    "player",
    "team",
    "season",
    "games",
    "points_per_game",
    "estimated_total_points"
]].head())
```

**7. Filter the Data**

```python
# Filter players averaging 20 or more points per game
high_scorers = df[df["points_per_game"] >= 20]

print("Players averaging 20 or more points per game:")
display(high_scorers)
```

**8. Unique values and categories**

```python
# Display unique player names
print("Unique players:")
print(df["player"].nunique())

# Display unique teams
print("\nUnique teams:")
print(df["team"].unique())

# Display unique positions
print("\nUnique positions:")
print(df["position"].unique())

# Count the number of records in each category
print("\nTeam counts:")
print(df["team"].value_counts())

print("\nPosition counts:")
print(df["position"].value_counts())
```

**9. Grouping and Aggregation**

```python
# Group players by position and calculate average statistics
position_summary = df.groupby("position")[
    ["points_per_game", "rebounds_per_game", "assists_per_game"]
].mean()

print("Average basketball statistics by position:")
display(position_summary)
```
